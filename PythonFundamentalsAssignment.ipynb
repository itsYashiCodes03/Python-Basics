# **TOPIC 3:OBJECTS,IDENTITY(is) vs EQUALITY(=)**

```python
#Q3.(1)
#  == checks if two objects have the same value, while is checks if they are the same object in memory
a = [1, 2, 3]
b = [1, 2, 3]

print(a is b)   # False

a = [1, 2, 3]
b = a

print(a == b)   # True
print(a is b)   # True
```

```python
#Q3.(2)
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```

```python
#Q3.(3)
x = 100
y = 100
print(x is y)
print(x == y)

p = 1000
q = 1000
print(p is q)
print(p == q)
```

# **TOPIC 7:FUNCIONS & PASSING ARGUMENTS**

```python
#Q7(1)
def my_abs(val):
  if val<0:
    return -val
  else:
    return val

print(my_abs(5))
print(my_abs(-5))
print(my_abs(0))


```

```python
#Q7(2)
#TypeError because Python cannot compare a string "Hi" with the number 0.
def my_abs(val):
    if val < 0:
        return -val
    else:
        return val

print(my_abs("Hi"))
```

```python
#Q7(3)
def print_abs(val):
    if val < 0:
        print(-val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
#The function prints 2.7 but does not use return, so x gets the default value None
```

```python
#Q7(4)
def inc_val(val):
  val=val+1
x=7
inc_val(x)
print(x)
#val is a separate local variable inside the function. Changing `val` does not change `x`; `x` remains `7`.


```

```python
#Q7(5)
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print(val1, val2)

x = 6
y = 3

swap_message(x, y)

print(x, y)
```

# Topic 8: Scope — Local vs Global Variables

```python
#Q8.(1)


#val is only defined as a parameter inside the function. print(val) is outside the function, so Python does not know what val is.
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(val)

```

```python
#Q8.(2)
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
#Q8.(3)

#Python treats counter inside the function as a local variable, but you try to use it before giving it a value.
counter=0
def increment_counter():
  counter=counter+1
increment_counter()
```

```python
#Q8(4)
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
